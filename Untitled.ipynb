# 1. What is AQI

<div style="font-family: Arial; padding:15px; border-radius:10px; background:#f4f6f7;">
  <h2 style="color:#2c3e50;">What is Air Quality Index (AQI)?</h2>
  <p>
    Air Quality Index (AQI) is a numerical scale used by governments to communicate air quality status to the public.
    It reflects how polluted the air is and the associated health risks.
  </p>
  <p>
    As pollutant concentration increases, AQI values rise, indicating worsening air quality and higher health risk levels.
  </p>
</div>
<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#eef2f7;">
  <h3 style="color:#34495e;">Indian (CPCB) AQI</h3>
  <p>
    AQI is calculated based on eight major pollutants:
  </p>
  <ul>
    <li>PM10 (Particulate Matter ≤ 10µm)</li>
    <li>PM2.5 (Particulate Matter ≤ 2.5µm)</li>
    <li>Carbon Monoxide (CO)</li>
    <li>Ozone (O₃)</li>
    <li>Nitrogen Dioxide (NO₂)</li>
    <li>Sulfur Dioxide (SO₂)</li>
    <li>Ammonia (NH₃)</li>
    <li>Lead (Pb)</li>
  </ul>
  <p>
    A minimum of three pollutants is required for AQI calculation, including at least one of PM10 or PM2.5.
  </p>
</div>
<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#ffffff; border:1px solid #ddd;">

<div style="font-family: Arial; padding:15px; border-radius:10px; background:#f5f7fa;">
  <h2 style="color:#2c3e50;">How to Calculate Air Quality Index (AQI)</h2>
  <p>
    The formula to calculate AQI is consistent with both Indian CPCB and US-EPA standards.
    AQI is computed separately for each pollutant, and the highest sub-index determines the final AQI.
  </p>
  <p>
    A sub-index is a linear function of pollutant concentration.
  </p>
</div>

<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#eef2f7;">
  <h3 style="color:#34495e;">AQI Formula</h3>
  <p style="font-size:18px; font-weight:bold; text-align:center;">
    Ip = [(IHi − ILo) / (BPHi − BPLo)] × (Cp − BPLo) + ILo
  </p>
</div>

<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#ffffff; border:1px solid #ddd;">
  <h3 style="color:#34495e;">Where:</h3>
  <ul>
    <li><b>Ip</b> = Index of pollutant p</li>
    <li><b>Cp</b> = Truncated concentration of pollutant p</li>
    <li><b>BPHi</b> = Upper concentration breakpoint (≥ Cp)</li>
    <li><b>BPLo</b> = Lower concentration breakpoint (≤ Cp)</li>
    <li><b>IHi</b> = AQI value corresponding to BPHi</li>
    <li><b>ILo</b> = AQI value corresponding to BPLo</li>
  </ul>
</div>

<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#fef9e7;">
  <h3 style="color:#b7950b;">Indian AQI Guidelines</h3>
  <ul>
    <li>A minimum of 3 parameters is required</li>
    <li>At least one must be PM10 or PM2.5</li>
    <li>16 hours of data is required for sub-index calculation</li>
  </ul>
</div>

<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#eafaf1;">
  <h3 style="color:#1e8449;">Example Calculation</h3>
  <p>
    Parameters: PM2.5, CO, Ozone
  </p>
  <p>
    Given: PM2.5 = 110 µg/m³
  </p>
  <ul>
    <li>BPHi = 120</li>
    <li>BPLo = 91</li>
    <li>IHi = 300</li>
    <li>ILo = 201</li>
  </ul>

  <p><b>Sub Index Calculation:</b></p>
  <p style="font-size:16px; font-weight:bold;">
    [(300 − 201) / (120 − 91)] × (110 − 91) + 201 = 265.86
  </p>

  <p>
    Repeat for other pollutants. The highest sub-index becomes the final AQI.
  </p>
</div>

```python
def sub_index(Cp, BLo, BHi, ILo, IHi):
    return ((IHi - ILo) / (BHi - BLo)) * (Cp - BLo) + ILo

# Breakpoints (simplified CPCB)
breakpoints = {
    'PM2_5': [(0,30,0,50),(31,60,51,100),(61,90,101,200),(91,120,201,300),(121,250,301,400),(251,500,401,500)],
    'PM10':  [(0,50,0,50),(51,100,51,100),(101,250,101,200),(251,350,201,300),(351,430,301,400),(431,600,401,500)],
    'NO2':   [(0,40,0,50),(41,80,51,100),(81,180,101,200),(181,280,201,300),(281,400,301,400),(401,1000,401,500)],
    'SO2':   [(0,40,0,50),(41,80,51,100),(81,380,101,200),(381,800,201,300),(801,1600,301,400),(1601,2000,401,500)]
}

def compute_subindex(value, pollutant):
    for (BLo, BHi, ILo, IHi) in breakpoints[pollutant]:
        if BLo <= value <= BHi:
            return sub_index(value, BLo, BHi, ILo, IHi)
    return None

def compute_aqi(row):
    sub_indices = []
    for p in ['PM2_5', 'PM10', 'NO2', 'SO2']:
        if pd.notna(row[p]):
            si = compute_subindex(row[p], p)
            if si is not None:
                sub_indices.append(si)
    return max(sub_indices) if sub_indices else None

aqi_df['AQI_reconstructed'] = aqi_df.apply(compute_aqi, axis=1)
```

```python
aqi_df['error'] = aqi_df['AQI'] - aqi_df['AQI_reconstructed']
aqi_df['abs_error'] = aqi_df['error'].abs()

# Summary metrics
print("MAE:", aqi_df['abs_error'].mean())
print("RMSE:", (aqi_df['error']**2).mean()**0.5)
print("Correlation:", aqi_df[['AQI','AQI_reconstructed']].corr().iloc[0,1])
```

```python
spatial_summary = pd.read_csv('spatial_summary.csv')

# Separate same vs cross district
same = spatial_summary[spatial_summary['Same_District'] == True]
cross = spatial_summary[spatial_summary['Same_District'] == False]

# Aggregate metrics
summary = (
    spatial_summary
    .groupby(['Pollutant', 'Same_District'])['Correlation']
    .agg(['mean', 'median', 'std', 'count'])
    .reset_index()
)

print(summary)
```

```python
same_mean = same.groupby('Pollutant')['Correlation'].mean()
cross_mean = cross.groupby('Pollutant')['Correlation'].mean()

spatial_lift = (same_mean - cross_mean).reset_index()
spatial_lift.columns = ['Pollutant', 'Spatial_Lift']

print(spatial_lift)
```

```python
def classify(row):
    same_corr = same_mean[row]
    cross_corr = cross_mean[row]
    lift = same_corr - cross_corr
    
    if same_corr >= 0.7 and lift >= 0.2:
        return 'Strong'
    elif same_corr >= 0.4:
        return 'Moderate'
    else:
        return 'Weak'

classification = pd.DataFrame({
    'Pollutant': same_mean.index,
    'Same_District_Corr': same_mean.values,
    'Cross_District_Corr': cross_mean.values,
    'Spatial_Lift': (same_mean - cross_mean).values
})

classification['Strength'] = classification['Pollutant'].apply(classify)

print(classification)
```

## 2) Temporal imputation (station-wise)

### 2.1 PM10

```python
def temporal_pm10(df):
    g = merged_df.copy()
    
    # 1. Linear interpolation (short gaps ≤ 2)
    g['PM10'] = g['PM10'].interpolate(method='linear', limit=2)
    
    # 2. Rolling median (medium gaps)
    rolling_med = g['PM10'].rolling(window=3, center=True, min_periods=1).median()
    mask = g['PM10'].isna()
    g.loc[mask, 'PM10'] = rolling_med[mask]
    
    # 3. Controlled forward/backward fill (≤ 2)
    g['PM10'] = g['PM10'].ffill(limit=2)
    g['PM10'] = g['PM10'].bfill(limit=2)
    
    return g

df = temporal_pm10(df)
```

```python
def temporal_secondary(df):
    g = df.copy()
    
    for col in ['PM2_5', 'NO2']:
        # 1. Linear interpolation (short gaps ≤ 2)
        g[col] = g[col].interpolate(method='linear', limit=2)
        
        # 2. Rolling median (only if enough non-NaN context exists)
        rolling_med = g[col].rolling(window=3, center=True, min_periods=2).median()
        
        # Fill only where rolling median is valid
        mask = g[col].isna() & rolling_med.notna()
        g.loc[mask, col] = rolling_med[mask]
    
    return g

df = temporal_secondary(df)
```

```python
def model_impute_secondary(df, target):
    g = df.copy()
    
    # --- Feature engineering (strictly past-dependent) ---
    g['PM10_lag1'] = g['PM10'].shift(1)
    g['PM10_lag2'] = g['PM10'].shift(2)
    g['PM10_roll3'] = g['PM10'].rolling(3).mean()
    
    features = ['PM10', 'PM10_lag1', 'PM10_lag2', 'PM10_roll3']
    
    # --- Train set ---
    train = g[g[target].notna()].dropna(subset=features)
    
    # --- Predict set ---
    test = g[g[target].isna()].dropna(subset=features)
    
    if len(train) < 20 or len(test) == 0:
        return g  # insufficient signal
    
    model = LGBMRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=4
    )
    
    model.fit(train[features], train[target])
    
    g.loc[test.index, target] = model.predict(test[features])
    
    return g

df = model_impute_secondary(df, 'PM2_5')
df = model_impute_secondary(df, 'NO2')
```

```python
def model_impute_so2(df):
    g = df.copy()
    
    # Feature engineering (reuse available signals)
    g['PM10_lag1'] = g['PM10'].shift(1)
    
    features = ['PM10', 'PM2_5', 'NO2', 'PM10_lag1']
    
    train = g[g['SO2'].notna()].dropna(subset=features)
    test = g[g['SO2'].isna()].dropna(subset=features)
    
    # Strict threshold due to weak signal
    if len(train) < 25 or len(test) == 0:
        return g
    
    model = LGBMRegressor(
        n_estimators=150,
        learning_rate=0.05,
        max_depth=3
    )
    
    model.fit(train[features], train['SO2'])
    
    g.loc[test.index, 'SO2'] = model.predict(test[features])
    
    return g

df = model_impute_so2(df)
```

```python
def model_impute_aqi(df):
    g = df.copy()
    
    # --- Feature engineering ---
    g['PM10_lag1'] = g['PM10'].shift(1)
    g['AQI_lag1'] = g['AQI'].shift(1)
    g['PM10_roll3'] = g['PM10'].rolling(3).mean()
    
    features = ['PM10', 'PM2_5', 'NO2', 'PM10_lag1', 'AQI_lag1', 'PM10_roll3']
    
    # --- Train ---
    train = g[g['AQI'].notna()].dropna(subset=features)
    
    # --- Predict ---
    test = g[g['AQI'].isna()].dropna(subset=features)
    
    if len(train) < 30 or len(test) == 0:
        return g
    
    model = LGBMRegressor(
        n_estimators=400,
        learning_rate=0.05,
        max_depth=5
    )
    
    model.fit(train[features], train['AQI'])
    
    g.loc[test.index, 'AQI'] = model.predict(test[features])
    
    return g

df = model_impute_aqi(df)
```

```python
def final_reconciliation(df):
    g = df.copy()
    
    pollutants = ['AQI', 'PM10', 'PM2_5', 'NO2', 'SO2']
    
    # 1. Physical bounds (non-negative)
    for col in pollutants:
        g[col] = g[col].clip(lower=0)
    
    # 2. Upper bound clipping (robust, prevents model spikes)
    for col in pollutants:
        q_low = g[col].quantile(0.01)
        q_high = g[col].quantile(0.99)
        
        iqr = q_high - q_low
        upper_cap = q_high + 1.5 * iqr
        
        g[col] = np.where(g[col] > upper_cap, upper_cap, g[col])
    
    return g

df = final_reconciliation(df)
```

```python
def fallback_imputation(df):
    g = df.copy()
    
    pollutants = ['PM10', 'PM2_5', 'NO2', 'SO2', 'AQI']
    
    # --- 1. Fix lag-induced NaNs (first rows) ---
    for col in ['PM10_lag1', 'PM10_lag2', 'AQI_lag1']:
        g[col] = g[col].bfill(limit=1)
    
    # --- 2. Local forward/backward fill (very strict) ---
    for col in pollutants:
        g[col] = g[col].ffill(limit=1)
        g[col] = g[col].bfill(limit=1)
    
    # --- 3. PM10-based fallback (only if still missing) ---
    # preserves structure since PM10 is dominant signal
    for col in ['PM2_5', 'NO2']:
        mask = g[col].isna() & g['PM10'].notna()
        g.loc[mask, col] = 0.4 * g.loc[mask, 'PM10']   # conservative scaling
    
    # --- 4. AQI fallback (last resort) ---
    mask = g['AQI'].isna() & g['PM10'].notna()
    g.loc[mask, 'AQI'] = 0.6 * g.loc[mask, 'PM10']
    
    return g

df = fallback_imputation(df)
```

```python
def knn_complete(df):
    g = df.copy()
    
    cols = ['AQI', 'PM10', 'PM2_5', 'NO2', 'SO2',
            'PM10_lag1', 'PM10_lag2', 'PM10_roll3', 'AQI_lag1']
    
    imputer = KNNImputer(
        n_neighbors=5,
        weights='distance',
        metric='nan_euclidean'
    )
    
    g[cols] = imputer.fit_transform(g[cols])
    
    return g

df_complete = knn_complete(df)
```
