# 1. What is AQI

<div style="font-family: Arial; padding:15px; border-radius:10px; background:#f4f6f7;">
  <h2 style="color:#2c3e50;">What is Air Quality Index (AQI)?</h2>
  <p>
    Air Quality Index (AQI) is a numerical scale used by governments to communicate air quality status to the public.
    It reflects how polluted the air is and the associated health risks.
  </p>
  <p>
    As pollutant concentration increases, AQI values rise, indicating worsening air quality and higher health risk levels.
  </p>
</div>
<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#eef2f7;">
  <h3 style="color:#34495e;">Indian (CPCB) AQI</h3>
  <p>
    AQI is calculated based on eight major pollutants:
  </p>
  <ul>
    <li>PM10 (Particulate Matter ≤ 10µm)</li>
    <li>PM2.5 (Particulate Matter ≤ 2.5µm)</li>
    <li>Carbon Monoxide (CO)</li>
    <li>Ozone (O₃)</li>
    <li>Nitrogen Dioxide (NO₂)</li>
    <li>Sulfur Dioxide (SO₂)</li>
    <li>Ammonia (NH₃)</li>
    <li>Lead (Pb)</li>
  </ul>
  <p>
    A minimum of three pollutants is required for AQI calculation, including at least one of PM10 or PM2.5.
  </p>
</div>

<div style="font-family: Arial; padding:15px; border-radius:10px; background:#f5f7fa;">
  <h2 style="color:#2c3e50;">How to Calculate Air Quality Index (AQI)</h2>
  <p>
    The formula to calculate AQI is consistent with both Indian CPCB and US-EPA standards.
    AQI is computed separately for each pollutant, and the highest sub-index determines the final AQI.
  </p>
  <p>
    A sub-index is a linear function of pollutant concentration.
  </p>
</div>

<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#eef2f7;">
  <h3 style="color:#34495e;">AQI Formula</h3>
  <p style="font-size:18px; font-weight:bold; text-align:center;">
    Ip = [(IHi − ILo) / (BPHi − BPLo)] × (Cp − BPLo) + ILo
  </p>
</div>

<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#ffffff; border:1px solid #ddd;">
  <h3 style="color:#34495e;">Where:</h3>
  <ul>
    <li><b>Ip</b> = Index of pollutant p</li>
    <li><b>Cp</b> = Truncated concentration of pollutant p</li>
    <li><b>BPHi</b> = Upper concentration breakpoint (≥ Cp)</li>
    <li><b>BPLo</b> = Lower concentration breakpoint (≤ Cp)</li>
    <li><b>IHi</b> = AQI value corresponding to BPHi</li>
    <li><b>ILo</b> = AQI value corresponding to BPLo</li>
  </ul>
</div>

<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#fef9e7;">
  <h3 style="color:#b7950b;">Indian AQI Guidelines</h3>
  <ul>
    <li>A minimum of 3 parameters is required</li>
    <li>At least one must be PM10 or PM2.5</li>
    <li>16 hours of data is required for sub-index calculation</li>
  </ul>
</div>

<div style="font-family: Arial; padding:15px; margin-top:10px; border-radius:10px; background:#eafaf1;">
  <h3 style="color:#1e8449;">Example Calculation</h3>
  <p>
    Parameters: PM2.5, CO, Ozone
  </p>
  <p>
    Given: PM2.5 = 110 µg/m³
  </p>
  <ul>
    <li>BPHi = 120</li>
    <li>BPLo = 91</li>
    <li>IHi = 300</li>
    <li>ILo = 201</li>
  </ul>

  <p><b>Sub Index Calculation:</b></p>
  <p style="font-size:16px; font-weight:bold;">
    [(300 − 201) / (120 − 91)] × (110 − 91) + 201 = 265.86
  </p>

  <p>
    Repeat for other pollutants. The highest sub-index becomes the final AQI.
  </p>
</div>

```python
def pre_imputation_integrity_checks(df, pollutant_cols=None):

    if pollutant_cols is None:
        pollutant_cols = ['AQI', 'PM10', 'PM2_5', 'NO2', 'SO2']

    df = df.copy()

    # # Ensure datetime
    # if pd.api.types.is_period_dtype(df['Date']):
    #     df['Date'] = df['Date'].dt.to_timestamp()
    # else:
    #     df['Date'] = pd.to_datetime(df['Date'])

    def process_station(station, g):
        g = g.sort_values('Date').copy()

        # Safe extraction (no dependency on column existence)
        # extracts region for the current station e.g for Anantnag (PM-10, PM-2.5) 1084
        # region will be Kashmir
        region = g['Region'].iloc[0] if 'Region' in g.columns else None 
        # district will be Anantnag
        district = g['District'].iloc[0] if 'District' in g.columns else None

        # --- Enforce continuous monthly index ---
        full_idx = pd.date_range(
            start=g['Date'].min(),
            end=g['Date'].max(),
            freq='MS'
        )

        g = g.set_index('Date').reindex(full_idx)
        g.index.name = 'Date'

        # Restore identifiers deterministically
        g['Station'] = station
        if region is not None:
            g['Region'] = region
        if district is not None:
            g['District'] = district

        # --- MAD-based outlier detection ---
        for col in pollutant_cols:
            if col not in g.columns:
                continue

            x = g[col]

            median = x.median(skipna=True)
            mad = np.median(np.abs(x - median))

            if mad == 0 or np.isnan(mad):
                continue

            robust_z = (x - median) / (1.4826 * mad)
            g.loc[np.abs(robust_z) > 4, col] = np.nan

        return g.reset_index()

    # Use (name, group) iteration → no KeyError possible
    result = []
    for station, g in df.groupby('Station'):
        result.append(process_station(station, g))

    return pd.concat(result, ignore_index=True)
```

```python
def temporal_imputation(df, pollutant_cols=None):
    """
    Station-wise temporal imputation with strict hierarchy:

    1) Linear interpolation for short gaps (≤2)
    2) Local rolling regression for medium gaps (3–5)
    3) Conditional ffill/bfill (≤2, low-variance only)

    No cross-station leakage.
    """

    if pollutant_cols is None:
        pollutant_cols = ['PM10', 'PM2_5', 'NO2', 'SO2']

    df = df.copy()

    df = df.sort_values(['Station', 'Date']).reset_index(drop=True)

    def _linear_short_gaps(g, col):
        return g[col].interpolate(method='linear', limit=2, limit_direction='both')

    def _identify_nan_runs(mask):
        # returns list of (start_idx, end_idx, length)
        runs = []
        start = None
        for i, val in enumerate(mask):
            if val and start is None:
                start = i
            elif not val and start is not None:
                runs.append((start, i - 1, i - start))
                start = None
        if start is not None:
            runs.append((start, len(mask) - 1, len(mask) - start))
        return runs

    def _rolling_regression_fill(g, col):
        values = g[col].values.copy()
        time_index = np.arange(len(g))

        nan_mask = np.isnan(values)
        runs = _identify_nan_runs(nan_mask)

        for start, end, length in runs:
            if 3 <= length <= 5:
                # window ±3
                left = max(0, start - 3)
                right = min(len(g), end + 4)

                idx = time_index[left:right]
                y = values[left:right]

                valid = ~np.isnan(y)
                if valid.sum() < 3:
                    continue

                X = idx[valid]
                Y = y[valid]

                # simple linear regression
                beta1, beta0 = np.polyfit(X, Y, 1)

                # fill only missing segment
                for t in range(start, end + 1):
                    values[t] = beta0 + beta1 * time_index[t]

        return pd.Series(values, index=g.index)

        
    def _conditional_propagation(g, col):
        x = g[col].copy()
        values = x.values.astype(float)
    
        station_std = np.nanstd(values)
        nan_mask = np.isnan(values)
        runs = _identify_nan_runs(nan_mask)
    
        for start, end, length in runs:
            if length > 2:
                continue
    
            left_idx = start - 1
            right_idx = end + 1
    
            left_val = values[left_idx] if left_idx >= 0 else np.nan
            right_val = values[right_idx] if right_idx < len(values) else np.nan
    
            if np.isnan(left_val) or np.isnan(right_val):
                continue
    
            local_std = np.std([left_val, right_val])
    
            if local_std >= station_std:
                continue
    
            # safe interpolation (not blind propagation)
            step = (right_val - left_val) / (length + 1)
            for i in range(length):
                values[start + i] = left_val + step * (i + 1)
    
        return pd.Series(values, index=g.index)

    def process_station(g):
        g = g.sort_values('Date').copy()

        for col in pollutant_cols:
            if col not in g.columns:
                continue

            # Step 1: linear interpolation (short gaps)
            g[col] = _linear_short_gaps(g, col)

            # Step 2: rolling regression (medium gaps)
            g[col] = _rolling_regression_fill(g, col)

            # Step 3: constrained propagation
            g[col] = _conditional_propagation(g, col)

        return g

    result = []
    for station, g in df.groupby('Station'):
        result.append(process_station(g))

    return pd.concat(result, ignore_index=True)
```

```python
def feature_based_imputation(df):
    """
    Feature-based imputation (strictly station-wise, no future leakage)

    1) PM2_5 ← PM10 (linear regression)
    2) NO2   ← PM10 (linear regression)
    3) SO2   ← stochastic sampling (station distribution)

    Assumes temporal imputation already applied.
    """

    df = df.copy()
    df = df.sort_values(['Station', 'Date']).reset_index(drop=True)

    def _fit_linear(x, y):
        # Simple OLS using numpy (robust, no sklearn dependency)
        beta1, beta0 = np.polyfit(x, y, 1)
        return beta0, beta1  # intercept, slope

    def process_station(g):
        g = g.sort_values('Date').copy()

        # ---------- PM2.5 and NO2 via PM10 ----------
        for target in ['PM2_5', 'NO2']:
            if target not in g.columns or 'PM10' not in g.columns:
                continue

            # training data (only observed rows)
            train_mask = g[target].notna() & g['PM10'].notna()

            if train_mask.sum() >= 5:
                X = g.loc[train_mask, 'PM10'].values
                Y = g.loc[train_mask, target].values

                intercept, slope = _fit_linear(X, Y)

                # predict only missing rows
                pred_mask = g[target].isna() & g['PM10'].notna()

                g.loc[pred_mask, target] = (
                    intercept + slope * g.loc[pred_mask, 'PM10']
                )

        # ---------- SO2 via stochastic sampling ----------
        if 'SO2' in g.columns:
            observed = g['SO2'].dropna()

            if len(observed) >= 5:
                mu = observed.mean()
                sigma = observed.std()

                if sigma > 0:
                    missing_idx = g['SO2'].isna()

                    n = missing_idx.sum()
                    if n > 0:
                        samples = np.random.normal(mu, sigma, size=n)

                        # truncate at >= 0
                        samples = np.clip(samples, 0, None)

                        g.loc[missing_idx, 'SO2'] = samples

        return g

    result = []
    for station, g in df.groupby('Station'):
        result.append(process_station(g))

    return pd.concat(result, ignore_index=True)
```

```python
def spatial_augmentation(df, pollutant_cols=None, min_peers=2):

    if pollutant_cols is None:
        pollutant_cols = ['PM10', 'PM2_5', 'NO2', 'SO2']

    df = df.copy()

    # ---------------------------
    # HARD STRUCTURE NORMALIZATION
    # ---------------------------

    # Always flatten index
    df = df.reset_index(drop=False)

    # Try to recover columns if renamed
    rename_map = {}
    for col in df.columns:
        c = str(col).lower()
        if c in ['district', 'level_0']:
            rename_map[col] = 'District'
        elif c in ['date', 'level_1']:
            rename_map[col] = 'Date'
        elif c in ['station', 'level_2']:
            rename_map[col] = 'Station'

    df = df.rename(columns=rename_map)

    # If still missing → cannot do spatial safely
    required = {'District', 'Date', 'Station'}
    if not required.issubset(df.columns):
        return df  # skip instead of crashing

    # Normalize Date
    if pd.api.types.is_period_dtype(df['Date']):
        df['Date'] = df['Date'].dt.to_timestamp()

    df = df.sort_values(['District', 'Date', 'Station']).reset_index(drop=True)

    # ---------------------------
    # SPATIAL IMPUTATION
    # ---------------------------

    def fill_spatial(x):
        # x.count() returns the number of non-NaN values
        if x.count() < min_peers:
            return x
        return x.fillna(x.median())

    for col in pollutant_cols:
        if col not in df.columns:
            continue

        # Use .transform() to safely update just the target column 
        # without changing the dataframe's structure
        df[col] = df.groupby(['District', 'Date'])[col].transform(fill_spatial)

    return df.drop(columns=['index'], errors='ignore')
def model_based_imputation(df, pollutant_cols=None, missing_threshold=0.4):

    if pollutant_cols is None:
        pollutant_cols = ['PM10', 'PM2_5', 'NO2', 'SO2']

    df = df.copy()

    # ----------- HARD NORMALIZATION (fix for your error) -----------
    # If Date is index → bring it back
    if 'Date' not in df.columns:
        df = df.reset_index()

    # If Station is index → bring it back
    if 'Station' not in df.columns:
        df = df.reset_index()

    # Final assertion (fail fast, not silently)
    if 'Date' not in df.columns or 'Station' not in df.columns:
        raise ValueError("DataFrame must contain 'Date' and 'Station' as columns")

    df['Date'] = pd.to_datetime(df['Date'])
    df = df.sort_values(['Station', 'Date']).reset_index(drop=True)

    # ----------- Time index (station-local only) -----------
    df['t_idx'] = df.groupby('Station').cumcount()

    # ----------- Lag features (strictly past) -----------
    for col in pollutant_cols:
        if col in df.columns:
            df[f'{col}_lag1'] = df.groupby('Station')[col].shift(1)
            df[f'{col}_lag2'] = df.groupby('Station')[col].shift(2)

    def build_features(g, target):
        feats = ['t_idx']

        if 'PM10' in g.columns:
            feats.append('PM10')

        for col in pollutant_cols:
            if col != target and col in g.columns:
                feats.append(col)

        feats += [f'{target}_lag1', f'{target}_lag2']

        return g[feats], feats

    def train_predict(g, target):
        g = g.copy()

        X, feat_cols = build_features(g, target)
        y = g[target]

        train_mask = y.notna() & X.notna().all(axis=1)
        pred_mask = y.isna() & X.notna().all(axis=1)

        if train_mask.sum() < 20:
            return g  # insufficient data

        model = lgb.LGBMRegressor(
            n_estimators=100,
            max_depth=5,
            learning_rate=0.05,
            min_child_samples=10,
            subsample=0.8,
            colsample_bytree=0.8,
            random_state=42
        )

        model.fit(X.loc[train_mask], y.loc[train_mask])

        if pred_mask.sum() > 0:
            preds = model.predict(X.loc[pred_mask])

            # variance preservation (critical)
            residuals = y.loc[train_mask] - model.predict(X.loc[train_mask])
            std = residuals.std()

            if std > 0:
                preds += np.random.normal(0, std, size=len(preds))

            g.loc[pred_mask, target] = preds

        return g

    result = []

    for station, g in df.groupby('Station'):
        g = g.copy()

        for target in pollutant_cols:
            if target not in g.columns:
                continue

            missing_ratio = g[target].isna().mean()

            if missing_ratio > missing_threshold:
                g_new = train_predict(g, target)

                # -------- district fallback -----------
                if g_new[target].isna().sum() > 0:
                    if 'District' in g.columns:
                        district = g['District'].iloc[0]
                        dfg = df[df['District'] == district].copy()

                        dfg = train_predict(dfg, target)

                        mask = g[target].isna()
                        g.loc[mask, target] = dfg.loc[g.index, target][mask]
                else:
                    g = g_new

        result.append(g)

    df_out = pd.concat(result, ignore_index=True)

    # cleanup
    drop_cols = [c for c in df_out.columns if '_lag' in c] + ['t_idx']
    df_out = df_out.drop(columns=drop_cols, errors='ignore')

    return df_out
def impute_aqi(df,POLLUTANTS):
    df = df.copy()
    result = []

    for station, g in df.groupby('Station'):
        g = g.sort_values('Date').copy()

        train_mask = g['AQI'].notna()
        pred_mask = g['AQI'].isna()

        if train_mask.sum() >= 10:
            X = g.loc[train_mask, POLLUTANTS]
            y = g.loc[train_mask, 'AQI']

            model = LinearRegression()
            model.fit(X, y)

            if pred_mask.sum() > 0:
                preds = model.predict(g.loc[pred_mask, POLLUTANTS])

                # variance preservation
                residuals = y - model.predict(X)
                std = residuals.std()

                if std > 0:
                    preds += np.random.normal(0, std, size=len(preds))

                g.loc[pred_mask, 'AQI'] = preds

        result.append(g)

    return pd.concat(result, ignore_index=True)
```

```python
def knn_impute_pollutants(df, cols=None, n_neighbors=5, weights='distance'):
    """
    Applies KNN imputation ONLY on selected pollutant columns.
    Other columns remain unchanged.

    Parameters:
    - df: input DataFrame
    - cols: columns to impute (default pollutants)
    - n_neighbors: K for KNN
    - weights: 'uniform' or 'distance'

    Returns:
    - DataFrame with imputed pollutant columns
    """

    if cols is None:
        cols = ['PM10','PM2_5','NO2','SO2']

    df = df.copy()

    # ----------------------------
    # SAFETY CHECKS
    # ----------------------------
    missing_cols = [c for c in cols if c not in df.columns]
    if missing_cols:
        raise ValueError(f"Columns not found: {missing_cols}")

    # Ensure numeric
    X = df[cols].astype(float)

    # ----------------------------
    # KNN IMPUTATION
    # ----------------------------
    imputer = KNNImputer(
        n_neighbors=n_neighbors,
        weights=weights,
        metric='nan_euclidean'
    )

    X_imputed = imputer.fit_transform(X)

    # ----------------------------
    # ASSIGN BACK
    # ----------------------------
    df[cols] = X_imputed

    return df
```

### Step 3. Reconstruct AQI using Indian AQI breakpoints (CPCB standard)

```python
def sub_index(Cp, BLo, BHi, ILo, IHi):
    return ((IHi - ILo) / (BHi - BLo)) * (Cp - BLo) + ILo

# Breakpoints (simplified CPCB)
breakpoints = {
    'PM2_5': [(0,30,0,50),(31,60,51,100),(61,90,101,200),(91,120,201,300),(121,250,301,400),(251,500,401,500)],
    'PM10':  [(0,50,0,50),(51,100,51,100),(101,250,101,200),(251,350,201,300),(351,430,301,400),(431,600,401,500)],
    'NO2':   [(0,40,0,50),(41,80,51,100),(81,180,101,200),(181,280,201,300),(281,400,301,400),(401,1000,401,500)],
    'SO2':   [(0,40,0,50),(41,80,51,100),(81,380,101,200),(381,800,201,300),(801,1600,301,400),(1601,2000,401,500)]
}

def compute_subindex(value, pollutant):
    for (BLo, BHi, ILo, IHi) in breakpoints[pollutant]:
        if BLo <= value <= BHi:
            return sub_index(value, BLo, BHi, ILo, IHi)
    return None

def compute_aqi(row):
    sub_indices = []
    for p in ['PM2_5', 'PM10', 'NO2', 'SO2']:
        if pd.notna(row[p]):
            si = compute_subindex(row[p], p)
            if si is not None:
                sub_indices.append(si)
    return max(sub_indices) if sub_indices else None

aqi_df['AQI_reconstructed'] = aqi_df.apply(compute_aqi, axis=1)
```

```python
import pandas as pd

def combine_air_quality_dfs(dfs):
    """
    Combines multiple heterogeneous air quality DataFrames.

    - Aligns schema (union of all columns)
    - Preserves all rows (no aggregation)
    - Fills missing columns with NaN
    - Standardizes dtypes

    Input:
        dfs: list of DataFrames

    Output:
        combined DataFrame
    """

    # ----------------------------
    # STEP 1: UNIFY COLUMN SPACE
    # ----------------------------
    all_columns = set()
    for df in dfs:
        all_columns.update(df.columns)

    all_columns = list(all_columns)

    aligned_dfs = []

    for df in dfs:
        df = df.copy()

        # Add missing columns
        missing_cols = set(all_columns) - set(df.columns)
        for col in missing_cols:
            df[col] = pd.NA

        # Ensure same column order
        df = df[all_columns]

        aligned_dfs.append(df)

    # ----------------------------
    # STEP 2: CONCAT (NO MERGE)
    # ----------------------------
    combined = pd.concat(aligned_dfs, ignore_index=True)

    # ----------------------------
    # STEP 3: TYPE NORMALIZATION
    # ----------------------------
    if 'Date' in combined.columns:
        combined['Date'] = pd.to_datetime(combined['Date'])

    # Force numeric pollutants
    for col in ['AQI','PM10','PM2_5','NO2','SO2']:
        if col in combined.columns:
            combined[col] = pd.to_numeric(combined[col], errors='coerce')

    # ----------------------------
    # STEP 4: SORT (CRITICAL)
    # ----------------------------
    sort_cols = [c for c in ['Region','District','Station','Date'] if c in combined.columns]
    combined = combined.sort_values(sort_cols).reset_index(drop=True)

    return combined
```

```python
from sklearn.preprocessing import LabelEncoder

def knn_impute_full(df, n_neighbors=5, weights='distance'):
    """
    KNN imputation on ALL columns (numeric + categorical)

    Steps:
    - Encode categorical variables
    - Convert Date to numeric
    - Apply KNN
    - Decode categorical back
    """

    df = df.copy()

    # ----------------------------
    # IDENTIFY COLUMN TYPES
    # ----------------------------
    cat_cols = ['Region', 'District', 'Station']
    num_cols = ['AQI','PM10','PM2_5','NO2','SO2']

    # Ensure columns exist
    cat_cols = [c for c in cat_cols if c in df.columns]
    num_cols = [c for c in num_cols if c in df.columns]

    # ----------------------------
    # ENCODE CATEGORICALS
    # ----------------------------
    encoders = {}
    for col in cat_cols:
        le = LabelEncoder()
        df[col] = df[col].astype(str)
        df[col] = le.fit_transform(df[col])
        encoders[col] = le

    # ----------------------------
    # HANDLE DATE (IMPORTANT)
    # ----------------------------
    if 'Date' in df.columns:
        df['Date'] = pd.to_datetime(df['Date'])
        df['Date'] = df['Date'].map(pd.Timestamp.toordinal)

    # ----------------------------
    # APPLY KNN
    # ----------------------------
    imputer = KNNImputer(
        n_neighbors=n_neighbors,
        weights=weights,
        metric='nan_euclidean'
    )

    imputed_array = imputer.fit_transform(df)

    df_imputed = pd.DataFrame(imputed_array, columns=df.columns)

    # ----------------------------
    # RESTORE DATE
    # ----------------------------
    if 'Date' in df_imputed.columns:
        df_imputed['Date'] = df_imputed['Date'].round().astype(int)
        df_imputed['Date'] = pd.to_datetime(
            df_imputed['Date'], origin='unix', unit='D', errors='coerce'
        )

    # ----------------------------
    # DECODE CATEGORICALS
    # ----------------------------
    for col, le in encoders.items():
        df_imputed[col] = df_imputed[col].round().astype(int)
        df_imputed[col] = le.inverse_transform(df_imputed[col])

    # ----------------------------
    # FINAL TYPE FIX
    # ----------------------------
    for col in num_cols:
        df_imputed[col] = pd.to_numeric(df_imputed[col], errors='coerce')

    return df_imputed
```

```python
def temporal_pm10(df):
    g = merged_df.copy()
    
    # 1. Linear interpolation (short gaps ≤ 2)
    g['PM10'] = g['PM10'].interpolate(method='linear', limit=2)
    
    # 2. Rolling median (medium gaps)
    rolling_med = g['PM10'].rolling(window=3, center=True, min_periods=1).median()
    mask = g['PM10'].isna()
    g.loc[mask, 'PM10'] = rolling_med[mask]
    
    # 3. Controlled forward/backward fill (≤ 2)
    g['PM10'] = g['PM10'].ffill(limit=2)
    g['PM10'] = g['PM10'].bfill(limit=2)
    
    return g

df = temporal_pm10(df)
```

```python
def model_impute_secondary(df, target):
    g = df.copy()
    
    # --- Feature engineering (strictly past-dependent) ---
    g['PM10_lag1'] = g['PM10'].shift(1)
    g['PM10_lag2'] = g['PM10'].shift(2)
    g['PM10_roll3'] = g['PM10'].rolling(3).mean()
    
    features = ['PM10', 'PM10_lag1', 'PM10_lag2', 'PM10_roll3']
    
    # --- Train set ---
    train = g[g[target].notna()].dropna(subset=features)
    
    # --- Predict set ---
    test = g[g[target].isna()].dropna(subset=features)
    
    if len(train) < 20 or len(test) == 0:
        return g  # insufficient signal
    
    model = LGBMRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=4
    )
    
    model.fit(train[features], train[target])
    
    g.loc[test.index, target] = model.predict(test[features])
    
    return g

df = model_impute_secondary(df, 'PM2_5')
df = model_impute_secondary(df, 'NO2')
```

```python
def model_impute_so2(df):
    g = df.copy()
    
    # Feature engineering (reuse available signals)
    g['PM10_lag1'] = g['PM10'].shift(1)
    
    features = ['PM10', 'PM2_5', 'NO2', 'PM10_lag1']
    
    train = g[g['SO2'].notna()].dropna(subset=features)
    test = g[g['SO2'].isna()].dropna(subset=features)
    
    # Strict threshold due to weak signal
    if len(train) < 25 or len(test) == 0:
        return g
    
    model = LGBMRegressor(
        n_estimators=150,
        learning_rate=0.05,
        max_depth=3
    )
    
    model.fit(train[features], train['SO2'])
    
    g.loc[test.index, 'SO2'] = model.predict(test[features])
    
    return g

df = model_impute_so2(df)
```

```python
def final_reconciliation(df):
    g = df.copy()
    
    pollutants = ['AQI', 'PM10', 'PM2_5', 'NO2', 'SO2']
    
    # 1. Physical bounds (non-negative)
    for col in pollutants:
        g[col] = g[col].clip(lower=0)
    
    # 2. Upper bound clipping (robust, prevents model spikes)
    for col in pollutants:
        q_low = g[col].quantile(0.01)
        q_high = g[col].quantile(0.99)
        
        iqr = q_high - q_low
        upper_cap = q_high + 1.5 * iqr
        
        g[col] = np.where(g[col] > upper_cap, upper_cap, g[col])
    
    return g

df = final_reconciliation(df)
```

```python
def fallback_imputation(df):
    g = df.copy()
    
    pollutants = ['PM10', 'PM2_5', 'NO2', 'SO2', 'AQI']
    
    # --- 1. Fix lag-induced NaNs (first rows) ---
    for col in ['PM10_lag1', 'PM10_lag2', 'AQI_lag1']:
        g[col] = g[col].bfill(limit=1)
    
    # --- 2. Local forward/backward fill (very strict) ---
    for col in pollutants:
        g[col] = g[col].ffill(limit=1)
        g[col] = g[col].bfill(limit=1)
    
    # --- 3. PM10-based fallback (only if still missing) ---
    # preserves structure since PM10 is dominant signal
    for col in ['PM2_5', 'NO2']:
        mask = g[col].isna() & g['PM10'].notna()
        g.loc[mask, col] = 0.4 * g.loc[mask, 'PM10']   # conservative scaling
    
    # --- 4. AQI fallback (last resort) ---
    mask = g['AQI'].isna() & g['PM10'].notna()
    g.loc[mask, 'AQI'] = 0.6 * g.loc[mask, 'PM10']
    
    return g

df = fallback_imputation(df)
```

```python
class JKAirQualityImputer:
    def __init__(self):
        # Base pollutants ordered by typical missingness & signal strength
        self.pollutants = ['PM10', 'PM2_5', 'NO2', 'SO2']
        self.aqi_model = RandomForestRegressor(n_estimators=100, min_samples_leaf=3, random_state=42)
        
        # MICE with ExtraTrees preserves non-linear relationships and maintains variance
        self.mice_imputer = IterativeImputer(
            estimator=ExtraTreesRegressor(n_estimators=50, min_samples_leaf=2, random_state=42),
            max_iter=10,
            imputation_order='roman', # Impute left-to-right based on column order
            random_state=42
        )

    def fit_transform(self, df):
        """
        Executes the full 6-step imputation pipeline.
        Expects a DataFrame with: Date, Region, District, Station, AQI, PM10, PM2_5, NO2, SO2
        """
        df = df.copy()
        
        # FIX: Convert PeriodDtype to standard Timestamp for pipeline compatibility
        if pd.api.types.is_period_dtype(df['Date']):
            df['Date'] = df['Date'].dt.to_timestamp()
        else:
            df['Date'] = pd.to_datetime(df['Date'])
            
        # Sort values strictly by Time per Station
        df = df.sort_values(['Station', 'Date']).reset_index(drop=True)
        
        # STEP 1: Bounding mathematically impossible values
        for col in self.pollutants + ['AQI']:
            df.loc[df[col] < 0, col] = np.nan
            
        # STEP 2: Micro-Temporal Interpolation (Strictly 1 month max)
        print("Executing Step 2: Micro-Temporal Interpolation...")
        for col in self.pollutants:
            df[col] = df.groupby('Station')[col].transform(
                lambda x: x.interpolate(method='linear', limit=1)
            )
            
        # STEP 3: Spatial Context Engineering (LOO District Median)
        print("Executing Step 3: Spatial Feature Engineering...")
        for col in self.pollutants:
            spatial_col = f'Spatial_{col}'
            
            # Calculate District Median excluding the current row's value (Leave-One-Out approximation)
            # For monthly data, median of the district for that specific month
            def loo_median(group):
                counts = group.notna().sum()
                if counts > 1:
                    # If other stations exist, use median of others. 
                    # (Approximated here by district median for performance)
                    return group.fillna(group.median())
                return group
                
            df[spatial_col] = df.groupby(['District', 'Date'])[col].transform(loo_median)
            # If district has no other sensors online, fill with global median temporarily 
            # (only used as a feature, not direct imputation)
            df[spatial_col] = df[spatial_col].fillna(df[col].median())

        # STEP 4: Stochastic Multivariate Model-Based Imputation
        print("Executing Step 4: Chained Equation Imputation (MICE)...")
        # Create an ordinal time feature for trend capture without forcing seasonality
        df['Time_Index'] = (df['Date'].dt.year - df['Date'].dt.year.min()) * 12 + df['Date'].dt.month
        
        spatial_cols = [f'Spatial_{p}' for p in self.pollutants]
        mice_features = ['Time_Index', 'AQI'] + spatial_cols + self.pollutants
        
        # We temporarily fill AQI with its median strictly for the pollutant imputation step.
        # This prevents MICE from dropping rows where AQI is missing. 
        # (We will properly reconstruct AQI in Step 5).
        temp_df = df[mice_features].copy()
        temp_df['AQI'] = temp_df['AQI'].fillna(temp_df['AQI'].median())
        
        # Reorder to ensure PM10 is imputed before PM2.5/NO2, before SO2
        imputed_matrix = self.mice_imputer.fit_transform(temp_df)
        imputed_df = pd.DataFrame(imputed_matrix, columns=mice_features)
        
        # Map imputed pollutants back to main dataframe
        for col in self.pollutants:
            df[col] = imputed_df[col]

        # STEP 5: Empirical AQI Reconstruction
        print("Executing Step 5: Empirical AQI Reconstruction...")
        # Train on rows where AQI is actually known
        known_aqi_mask = df['AQI'].notna()
        missing_aqi_mask = df['AQI'].isna()
        
        if missing_aqi_mask.sum() > 0:
            X_aqi = df.loc[known_aqi_mask, self.pollutants]
            y_aqi = df.loc[known_aqi_mask, 'AQI']
            
            self.aqi_model.fit(X_aqi, y_aqi)
            
            # Predict only for the missing values
            X_missing = df.loc[missing_aqi_mask, self.pollutants]
            df.loc[missing_aqi_mask, 'AQI'] = self.aqi_model.predict(X_missing)

        # STEP 6: Edge Case Resolution (Empirical Sampling for SO2/sparse features)
        print("Executing Step 6: Edge Case Resolution...")
        for col in self.pollutants:
            if df[col].isna().sum() > 0:
                # If any NaN slipped through (e.g., entire district missing for months),
                # sample from the station's own historical empirical distribution
                nan_mask = df[col].isna()
                station_emp_dist = df.groupby('Station')[col].apply(lambda x: x.dropna().values)
                
                for idx in df[nan_mask].index:
                    station = df.loc[idx, 'Station']
                    vals = station_emp_dist.get(station, [])
                    if len(vals) > 0:
                        df.loc[idx, col] = np.random.choice(vals)
                    else:
                        # Absolute fallback: District distribution sampling
                        df.loc[idx, col] = np.random.choice(df[col].dropna().values)

        # Cleanup: Drop auxiliary spatial features before returning
        df = df.drop(columns=spatial_cols + ['Time_Index'])
        print("Imputation Complete!")
        return df
```
